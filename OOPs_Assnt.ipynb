1.   In Python, the built‑in open() function is used to open a file and returns a file object for reading, writing, or appending.


**File Opening Modes :**
The different modes of opening a file in Python are as follows:
  - Read Mode ('r')
  - Write Mode ('w')
  - Append Mode ('a')
  - Exclusive Mode ('x')
  - Binary Write Mode ('wb')


  **Explain each mode of file opening python**

  - Read (r): Opens a file for reading only. If the file does not exist, it raises a FileNotFoundError. Example: with open('example.txt', 'r') as file: content = file.read()

  - Write (w): Opens a file for writing. If the file already exists, its content will be overwritten. If the file does not exist, a new file will be created. Example: with open('example.txt', 'w') as file: file.write('Hello, world!')

  - Append (a): Opens a file for appending. If the file does not exist, it will create a new file. If the file already exists, it will be positioned at the end. Example: with open('example.txt', 'a') as file: file.write('Hello, world!')

  - Write and Append (w+): Opens a file for both writing and appending. If the file already exists, it will be truncated to zero length. Example: with open('example.txt', 'w+') as file: file.write('Hello, world!')

  - Binary (b): Opens a file for reading or writing binary data. This mode is useful for non-text files such as images, audio, or binary data. Example: with open('image.jpg', 'rb') as file: binary_data = file.read()


2.   The close() function is used to ensure that all data is written to disk and that the file is properly closed. It is particularly important for output files, as they are often buffered, meaning that data is temporarily stored in memory before being written to disk.

 If a file is not closed, the data may not be saved to disk, and the file may become inaccessible. Closing a file also releases system resources, preventing the system from wasting memory or processing power. It is a good programming practice to call close() when writing to a file to ensure that all changes are saved and that the file is properly closed.


```python
# 3

file = open('test.txt','w')
file.write('I want to become a Data Scientist')
file.close()

# Reopen the file in read mode to read its content
file = open('test.txt', 'r')
content = file.read()
print(content)
file.close()
```

5.   **Explain why with statement is used with open():**
The with statement in Python is used with open() to manage file operations efficiently. It ensures that the file is automatically closed after the block of code is completed, regardless of whether an exception occurs.

Using the with statement together with open() for file handling is considered best practice because it provides automatic resource management and makes your code safer and cleaner.

**Advantages:**

- Automatic File Closing:  

  - When you use with open(...) as f:, Python automatically closes the file when the block ends — even if an exception occurs.
  - Without with, you must manually call f.close(), and if you forget, it can cause memory leaks or file locks.
- Exception Safety:

  - If an error happens inside the with block, the file is still closed properly.
  - This prevents issues like corrupted files or incomplete writes.

- Cleaner Syntax:

  - No need for try-finally blocks to ensure closing.
  - Code is shorter and easier to read.
- Resource Efficiency:

  - Frees up file handles immediately after use, which is important when working with many files or limited system resources.

6.   **Write():** This function is used to write a single string to a file. For instance, if you have a string variable message and want to write it to a file named example.txt.

This will create or overwrite the example.txt file with the string Hello, Python!. The write() function does not add a newline character at the end of the string, so you would need to add one manually if you want to write each string on a new line.

**Writelines():** This function is used to write a list of strings to a file. For example, if you have a list of lines and want to write them to a file.

This will append each line from the lines list to the example.txt file without overwriting the existing content. The writelines() function automatically adds newline characters between the strings, making it easier to write multiple lines on a new line.

```python
# 6
## write()
# Writing single strings
with open("example_write.txt", "w") as file:
    file.write("Hello World!\n")
    file.write("Welcome to Python File Handling.")
    file.close()

file1 = open("example_write.txt", "r")
print(file1.read())
```

```python
# 6
## writelines()

lines_list = [
    "First line of text.\n",
    "Second line of text.\n",
    "Third line of text.\n"
]

# Writing a list of strings
with open("example_writelines.txt", "w") as file:
    file.writelines(lines_list)
    file.close()

file2 = open("example_writelines.txt", "r")
print(file2.read())
```
